```python
# Import necessary libraries
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

# Define the ResNet18 building block
def resnet_block(input_layer, filters, kernel_size=3, stride=1, conv_shortcut=True, name=None):
    x = input_layer

    # Shortcut connection
    shortcut = x
    if conv_shortcut:
        shortcut = layers.Conv2D(filters, 1, strides=stride, name=name + '_0_conv')(x)  # Match the number of filters
        shortcut = layers.BatchNormalization(epsilon=1.001e-5, name=name + '_0_bn')(shortcut)

    # Residual path
    x = layers.Conv2D(filters, kernel_size, padding='SAME', strides=stride, name=name + '_1_conv')(x)
    x = layers.BatchNormalization(epsilon=1.001e-5, name=name + '_1_bn')(x)
    x = layers.Activation('relu', name=name + '_1_relu')(x)

    x = layers.Conv2D(filters, kernel_size, padding='SAME', name=name + '_2_conv')(x)
    x = layers.BatchNormalization(epsilon=1.001e-5, name=name + '_2_bn')(x)

    x = layers.Add(name=name + '_add')([shortcut, x])
    x = layers.Activation('relu', name=name + '_out')(x)

    return x

# Define the custom ResNet18 model
def build_resnet18(input_shape, num_classes):
    inputs = keras.Input(shape=input_shape)

    # Initial convolution and max-pooling
    x = layers.Conv2D(64, 7, strides=2, padding='SAME', name='conv1_conv')(inputs)
    x = layers.BatchNormalization(epsilon=1.001e-5, name='conv1_bn')(x)
    x = layers.Activation('relu', name='conv1_relu')(x)
    x = layers.MaxPooling2D(3, strides=2, name='pool1_pool')(x)

    # Building residual blocks
    x = resnet_block(x, 64, name='conv2_block1')
    x = resnet_block(x, 64, conv_shortcut=False, name='conv2_block2')

    x = resnet_block(x, 128, stride=2, name='conv3_block1')
    x = resnet_block(x, 128, conv_shortcut=False, name='conv3_block2')

    x = resnet_block(x, 256, stride=2, name='conv4_block1')
    x = resnet_block(x, 256, conv_shortcut=False, name='conv4_block2')

    x = resnet_block(x, 512, stride=2, name='conv5_block1')
    x = resnet_block(x, 512, conv_shortcut=False, name='conv5_block2')

    # Global Average Pooling and Output Layer
    x = layers.GlobalAveragePooling2D(name='avg_pool')(x)
    outputs = layers.Dense(num_classes, activation='softmax', name='predictions')(x)

    # Instantiate model
    model = keras.Model(inputs, outputs, name='ResNet18')
    return model

# Define paths for your dataset
train_dataset_path = "/home/user/JAYANT/training_3"
test_dataset_path = "/home/user/JAYANT/testing_3"

# Data Generators
testGen = ImageDataGenerator()
trainGen = ImageDataGenerator(validation_split=0.15, vertical_flip=True, horizontal_flip=True, rotation_range=0.2)

# Data loaders
trainGenerator = trainGen.flow_from_directory(train_dataset_path, subset="training", target_size=(224, 224), batch_size=128)
validationGenerator = trainGen.flow_from_directory(train_dataset_path, subset="validation", target_size=(224, 224), batch_size=128)
testGenerator = testGen.flow_from_directory(test_dataset_path, target_size=(224, 224), shuffle=False)

# Create the ResNet18 model
model = build_resnet18(input_shape=(224, 224, 3), num_classes=2)
model.summary()

# Compile the model
model.compile(optimizer=Adam(learning_rate=0.0001), loss='categorical_crossentropy', metrics=['accuracy'])

# Train the model
r = model.fit(
  trainGenerator,
  validation_data=validationGenerator,
  epochs=70,
  steps_per_epoch=len(trainGenerator),
  validation_steps=len(validationGenerator)
)

# Plotting loss
plt.plot(r.history['loss'], label='train loss')
plt.plot(r.history['val_loss'], label='val loss')
plt.legend()
plt.show()
plt.savefig('LossVal_loss.png')

# Plotting accuracy
plt.plot(r.history['accuracy'], label='train acc')
plt.plot(r.history['val_accuracy'], label='val acc')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
plt.savefig('AccVal_acc.png')

# Making predictions
CLASS_NAMES = list(trainGenerator.class_indices.keys())
predictions = np.argmax(model.predict(testGenerator), axis=1)

# Accuracy and classification report
acc = accuracy_score(testGenerator.labels, predictions)
cm = tf.math.confusion_matrix(testGenerator.labels, predictions)
clr = classification_report(testGenerator.labels, predictions, target_names=CLASS_NAMES)

print("Test Accuracy: {:.3f} %".format(acc * 100))

# Confusion matrix
plt.figure(figsize=(8, 8))
sns.heatmap(cm, annot=True, fmt='g', vmin=0, cmap='Reds', cbar=False)
plt.xticks(ticks=np.arange(2) + 2, labels=CLASS_NAMES)
plt.yticks(ticks=np.arange(2) + 2, labels=CLASS_NAMES)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()

print("Classification Report:\n-----------------------------\n", clr)
```
